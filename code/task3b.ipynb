```python
import cv2
import numpy as np

""" Simple script that shows a reference frame on the left and enables the user to find a matching frame on the right after one rotation by going through the video"""

VIDEO_PATH = "AO_Rotation_Recording_full3.mp4"
REFERENCE_FRAME_INDEX = 20 

def main():
    cap = cv2.VideoCapture(VIDEO_PATH)
    fps = cap.get(cv2.CAP_PROP_FPS)

    # reference frame
    cap.set(cv2.CAP_PROP_POS_FRAMES, REFERENCE_FRAME_INDEX)
    ret, reference_frame = cap.read()

    # Initialize the current frame index
    current_frame_index = REFERENCE_FRAME_INDEX
    current_frame = reference_frame.copy()

    # Create a window
    cv2.namedWindow('Frame Comparison', cv2.WINDOW_NORMAL)

    # Function to update the displayed frames
    def update_frames():
        nonlocal current_frame, current_frame_index
        cap.set(cv2.CAP_PROP_POS_FRAMES, current_frame_index)
        ret, current_frame = cap.read()
        if not ret:
            print(f"Error: Could not read frame at index {current_frame_index}.")
            return

        # Calculate the time difference
        time_diff = (current_frame_index - REFERENCE_FRAME_INDEX) / fps

        cv2.imshow('Frame Comparison', np.hstack((reference_frame, current_frame)))
        print(f"Current Frame Index: {current_frame_index}, Time Difference: {time_diff:.2f} seconds")

    update_frames()

    while True:
        # Display the frames
        key = cv2.waitKey(1) & 0xFF

        # Handle key presses
        if key == ord('q'):
            break
        elif key == ord('n'):
            current_frame_index += 1
            update_frames()
        elif key == ord('p'):
            current_frame_index -= 1
            update_frames()
        elif key == ord('s'):
            current_frame_index += 50
            update_frames()

    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```
